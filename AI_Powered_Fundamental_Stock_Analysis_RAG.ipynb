```python
# Import the library used to retrieve financial data
import yfinance as yf

# Select Microsoft using its ticker symbol
msft = yf.Ticker("MSFT")
```

```python
# Retrieve Microsoft's company information
company_info = msft.info

# Display the company name and industry
print("Company Name:", company_info.get("longName"))
print("Industy:", company_info.get("industry"))
print("Sector:", company_info.get("sector"))
```

```python
# Retrieve Microsoft's annual income statement
income_statement = msft.incomestmt

# Display the available financial data
display(income_statement)
```

```python
# Retrieve Microsoft's annual balance sheet
balance_sheet = msft.balance_sheet

# Display the available financial data
display(balance_sheet)
```

```python
# Retrieve Microsoft's annual cash flow statement
cash_flow =msft.cashflow

# Display the available financial data
display(cash_flow)
```

```python
# Retrieve Microsoft's historical stock prices for the last five years
historical_prices = msft.history(period="5y")

# Display the first five rows
display(historical_prices.head())
```

```python
# Retrieve selected financial metrics from the company information
financial_metrices = {
    "EPS": company_info.get("trailingEps"),
    "P/E Ratio": company_info.get("trailingPD"),
    "Market Capitalization": company_info.get("marketcap")
}

# Dispaly the selected financial metrices
print(financial_metrices)
```

```python
# Convert company information into readable text
company_text = "\n".join(
    f"{key}: {value}"
    for key, value in company_info.items()
)

# convert the income statement into readable text
income_text = income_statement.to_string()

# convert the balance sheet into readable text
balance_text = balance_sheet.to_string()

# convert the cash flow statement into readable text
cash_flow_text = cash_flow.to_string()

# convert financial metrices into readable text
metrices_text = "\n".join(
     f"{key}: {value}"
     for key, value in financial_metrices.items()
 )

print("Company information text prepared:", len(company_text), "characters")
print("Income statement text prepared:", len(income_text), "characters")
print("Balance sheet text prepared:", len(balance_text), "characters")
print("Cash flow text prepared:", len(cash_flow_text), "characters")
print("Financial metrices text prepared:", len(metrices_text), "characters")
```

```python
# Combine the collected financial information into one document
financial_document = f"""
MICROSOFT COMPANY INFORMATION
{company_text}

MICROSOFT INCOME STATEMENT
{income_text}

MICROSOFT BALANCE SHEET
{balance_text}

MICROSOFT CASH FLOW STATEMENT
{cash_flow_text}

MICROSOFT FINANCIAL METRICES
{metrices_text}
"""
print("Financial document created successfully.")
print("Total characters:", len(financial_document))
print("\nPreview of the document:")
print(financial_document[:500])
```

```python
# Split the financial document into chunks of 1000 characters
# Keep 200 characters overlapping between consecutive chunks

chunk_size = 1000
overlap = 200

text_chunks = []

for start in range(0, len(financial_document), chunk_size - overlap):
    text_chunks.append(financial_document[start:start + chunk_size])

print("Document length:", len(financial_document))
print("Total text chunks:", len(text_chunks))
print("First chunk length:", len(text_chunks[0]))
print("Last chunk length:", len(text_chunks[-1]))
```

```python
print("Document length:", len(financial_document))
print("Chunk count:", len(text_chunks))
print("Chunk lengths:", [len(chunk) for chunk in text_chunks[:5]])
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding Model Loaded Successfully!")
```

```python
# Convert each text chunk into a numerical embedding
embeddings = embedding_model.encode(text_chunks)

print("Number of embeddings:", len(embeddings))
print("Embedding dimensions:", embeddings.shape)
```

```python
import faiss
import numpy as np

# convert embeddings to the format FAISS requires
embeddings = np.array(embeddings, dtype="float32")

# create the FAISS index
index = faiss.IndexFlatL2(embeddings.shape[1])

# add all embeddings to the index
index.add(embeddings)

print("FAISS index created successfully!")
print("Total vectors indexed:", index.ntotal)
```

```python
# Search for text chunks related to Microsoft's revenue
query = "Microsoft total revenue for fiscal year 2026"

# Convert the question into an embedding
query_embedding = embedding_model.encode([query])
query_embedding = np.array(query_embedding, dtype="float32")

# Retrieve the 3 closest chunks
distances, indices = index.search(query_embedding, k=3)

print("Search query:", query)
print("\nRelevent text chunks:\n")

for i in indices[0]:
  print(text_chunks[i][:300])
  print("-" * 50)
```

```python
def retrieve_context(question, k=8):
  # convert the question into an embedding
  query_embedding = embedding_model.encode([question])
  query_embedding = np.array(query_embedding, dtype="float32")

  # Find the 8 most relevent text chunks
  distances, indices = index.search(query_embedding, k)

  # Combine the retrieved chunks into one context
  context = "\n\n".join(
      text_chunks[i]
      for i in indices[0]
      if i != -1
  )

  return context

print("Retrievel function created successfully!")
```

```python
def ask_financial_question(question):
  # Retrieve relevent information from the financial document
  context = retrieve_context(question, k=8)

  # Ask Gemini to answer using the retrieved information
  prompt = f"""
  Answer the question only using the financial context below.
  If the answer is not available in the context, say so.

  Financial context:
  {context}

  Question: {question}
  """

  response = client.models.generate_content(
      model="gemini-3.5-flash",
      contents=prompt
  )

  return response.text

print("Financial question-answering function created successfully!")
```

```python
question = "What was Microsoft's total revenue in fiscal year 2026?"

answer = ask_financial_question(question)

print("Question:", question)
print("\nAnswer:", answer)
```
